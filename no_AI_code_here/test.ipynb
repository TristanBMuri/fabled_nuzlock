```python
from poke_env import RandomPlayer, ServerConfiguration
import time

starting_port = 8020
server_id = 0

server = ServerConfiguration(
        websocket_url=f"ws://localhost:{starting_port + server_id}/showdown/websocket",
        authentication_url=f"http://localhost:{starting_port + server_id}/action.php",
    )
```

```python
player_1 = RandomPlayer(server_configuration=server, battle_format="gen8randombattle")
player_2 = RandomPlayer(server_configuration=server, battle_format="gen8randombattle")

start = time.time()
result = await player_1.battle_against(player_2, n_battles=50)
end = time.time()

print(f"Time to run 50 battles, one at a time: {end - start:.2f} seconds")
print(result)
```
